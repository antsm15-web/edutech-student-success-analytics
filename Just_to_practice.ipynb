```python
class Binaria:
    def __init__(self, nombre, periodo):
        self.nombre = nombre
        self.periodo = periodo

    def mostrar(self):
        print(f"{self.nombre}: {self.periodo}")
```

```python
hs0705 = Binaria("HS 0705+6700", 0.09564665)

hs0705.mostrar()
```

```python
def pedir_participante():
    nombre = input("Nombre: ")
    edad = int(input("Edad: "))
    correo = input("Correo: ")

    return {
        "nombre": nombre,
        "edad": edad,
        "correo": correo
    }

participantes = []

while True:
    participantes.append(pedir_participante())

    if input("¿Agregar otro? (s/n): ").lower() != "s":
        break
```
